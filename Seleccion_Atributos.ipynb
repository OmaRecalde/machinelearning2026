```python
import numpy as np
from sklearn.feature_selection import VarianceThreshold
```

```python
# Carga de datos
X = [
    [0, 0, 1],
    [0, 1, 0],
    [1, 0, 0],
    [0, 1, 1],
    [0, 1, 0],
    [0, 1, 1]
]

print(X)
```

```python
# Limpieza de datos: selección de atributos
# Fijamos un umbral (th) para decidir qué varianza mínima deben tener los valores
# de una columna para considerarse relevante a mantener en el estudio
th = 0.2
sel = VarianceThreshold(threshold=th)

# Ajustar y transformar X eliminando las columnas con varianza <= th
X_transformed = sel.fit_transform(X)

print(X_transformed)
```
