```python
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
# Carga de datos de entrenamiento con valores ausentes (np.nan)
X_train = [
    [1, 4],
    [np.nan, 3],
    [7, 8]
]
```

```python
# Limpieza de datos: imputación de valores ausentes usando la media
imputer = SimpleImputer(strategy='mean')

# Ajustar y transformar X_train calculando la media de cada columna
X_train_clean = imputer.fit_transform(X_train)

print(X_train_clean)
```

```python
# Carga de datos de test
X_test = [
    [np.nan, 10],
    [6, np.nan],
    [8, 2]
]

# Imputar datos de test reutilizando las medias aprendidas de X_train (evita data leakage)
X_test_clean = imputer.transform(X_test)

print(X_test_clean)
```
