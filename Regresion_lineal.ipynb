```python
# Importamos las librerías
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
# Creamos el conjunto de datos
df = pd.DataFrame({
    "horas": [1, 2, 3, 4, 5],
    "nota": [2, 3, 5, 4, 6]
})

X = df[["horas"]]
y = df[["nota"]]

print(X)
```

```python
# Creamos y entrenamos nuestro modelo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
# Imprimimos el intercepto y el coeficiente (pendiente)
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
# Realizamos la nueva predicción (usando DataFrame para mantener el nombre de la característica)
nueva_hora = pd.DataFrame({"horas": [6]})
prediccion = modelo.predict(nueva_hora)

print("Predicción:", prediccion)
```

```python
# Creamos los puntos de la recta
X_linea = np.linspace(1, 6, 100)
X_linea_df = pd.DataFrame({"horas": X_linea})
y_linea = modelo.predict(X_linea_df)

# Graficamos los datos
plt.figure(figsize=(8, 5))

# 1. Puntos originales del dataset
plt.scatter(X, y, color="blue", label="Datos reales")

# 2. Recta de regresión lineal
plt.plot(X_linea, y_linea, color="red", label="Recta de regresión")

# 3. Punto predicho (6 horas -> 6.7 nota)
plt.scatter([6], prediccion, color="green", s=100, zorder=5, label=f"Predicción (6h, {prediccion[0][0]:.1f})")

plt.title("Regresión Lineal: Horas de estudio vs Nota")
plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.legend()
plt.grid(True)
plt.show()
```
